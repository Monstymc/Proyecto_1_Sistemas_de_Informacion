## Fase 2: Definición de KPIs

| # | KPI | Fórmula | Stakeholder |
|---|-----|---------|-------------|
| 1 | Ingresos Totales Anuales | Σ ventas.total | Gerente General |
| 2 | Ticket Promedio | Ingresos Totales / N° Transacciones | Gerente de Ventas |
| 3 | Unidades Vendidas | Σ ventas.cantidad | Gerente de Inventario |
| 4 | Top 5 Productos (ingresos) | GROUP BY producto → SUM(total) | Gerente de Ventas |
| 5 | Top 5 Clientes (ingresos) | GROUP BY cliente → SUM(total) | Gerente de Marketing |
| 6 | Ingresos por Categoría | GROUP BY categoria → SUM(total) | Gerente General |
| 7 | Ingresos Mensuales | GROUP BY mes → SUM(total) | Gerente General |
| 8 | Ingresos por Día de Semana | GROUP BY dayofweek → SUM(total) | Gerente de Marketing |

## Fase 3: Implementación del Dashboard

### 3.1 Importar librerías

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import matplotlib.patches as mpatches
import matplotlib.ticker as ticker
import warnings
warnings.filterwarnings('ignore')

# Estilo visual profesional
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor']   = '#F8F9FA'
plt.rcParams['axes.grid']        = True
plt.rcParams['grid.alpha']       = 0.35
plt.rcParams['grid.linestyle']   = '--'
plt.rcParams['font.family']      = 'DejaVu Sans'
plt.rcParams['axes.spines.top']   = False
plt.rcParams['axes.spines.right'] = False

COLORES = ['#2D6A9F','#E07B39','#3BAB6F','#C0392B','#8E44AD',
           '#16A085','#D4AC0D','#1ABC9C','#E74C3C','#2980B9']
CP = '#2D6A9F'  # color primario
CA = '#E07B39'  # color acento

print(' Librerías cargadas correctamente')
```

### 3.3 Cálculo de KPIs

```python
# KPI 1 — Ingresos Totales Anuales
ingresos_totales = ventas['total'].sum()

# KPI 2 — Ticket Promedio
ticket_promedio = ingresos_totales / ventas['idVenta'].nunique()

# KPI 3 — Unidades Vendidas
unidades_vendidas = int(ventas['cantidad'].sum())

# KPI 4 — Top 5 Productos por ingresos
top_productos = (ventas.groupby('nombreProducto')['total']
                 .sum().sort_values(ascending=False).head(5))

# KPI 5 — Top 5 Clientes por ingresos
top_clientes = (ventas.groupby('nombreUsuario')['total']
                .sum().sort_values(ascending=False).head(5))

# KPI 6 — Ingresos por Categoría
ing_categoria = (ventas.groupby('categoria')['total']
                 .sum().sort_values(ascending=False))

# KPI 7 — Ingresos Mensuales
ventas_mensuales = ventas.groupby('mes')['total'].sum()
ventas_mensuales.index = [str(p) for p in ventas_mensuales.index]
meses_labels = ['Ene','Feb','Mar','Abr','May','Jun','Jul','Ago','Sep','Oct','Nov','Dic']

# KPI 8 — Ingresos por Día de Semana
orden_dias = ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']
nom_dias   = {'Monday':'Lunes','Tuesday':'Martes','Wednesday':'Miércoles',
              'Thursday':'Jueves','Friday':'Viernes','Saturday':'Sábado','Sunday':'Domingo'}
ventas_semana = (ventas.groupby('dia_semana')['total']
                 .sum().reindex(orden_dias).fillna(0))
ventas_semana.index = [nom_dias[d] for d in ventas_semana.index]

print(f'KPI 1 — Ingresos Totales   : $ {ingresos_totales:,.2f}')
print(f'KPI 2 — Ticket Promedio    : $ {ticket_promedio:.2f}')
print(f'KPI 3 — Unidades Vendidas  : {unidades_vendidas:,}')
print()
print('KPI 4 — Top 5 Productos:')
print(top_productos.to_string())
print()
print('KPI 5 — Top 5 Clientes:')
print(top_clientes.to_string())
print()
print('KPI 7 — Ingresos Mensuales:')
print(ventas_mensuales.to_string())
```

### 3.4 Dashboard 1 — Reporte Ejecutivo Anual

```python
fig = plt.figure(figsize=(18, 11))
fig.suptitle('Virvuto — Dashboard Ejecutivo  |  Año 2025 (12 meses)',
             fontsize=20, fontweight='bold', color='#1A1A2E', y=0.99)
gs = gridspec.GridSpec(3, 4, figure=fig, hspace=0.6, wspace=0.42)

def tarjeta(ax, titulo, valor, sub, color):
    ax.set_facecolor(color)
    for sp in ax.spines.values(): sp.set_visible(False)
    ax.set_xticks([]); ax.set_yticks([])
    ax.text(0.5,0.72,titulo, transform=ax.transAxes, ha='center', fontsize=11, color='white', alpha=0.88)
    ax.text(0.5,0.40,valor,  transform=ax.transAxes, ha='center', fontsize=21, fontweight='bold', color='white')
    ax.text(0.5,0.12,sub,    transform=ax.transAxes, ha='center', fontsize=9,  color='white', alpha=0.72)

tarjeta(fig.add_subplot(gs[0,0]), 'Ingresos Totales',  f'${ingresos_totales:,.0f}', 'Año 2025',             '#2D6A9F')
tarjeta(fig.add_subplot(gs[0,1]), 'Ticket Promedio',   f'${ticket_promedio:.2f}',   'Por transacción',      '#E07B39')
tarjeta(fig.add_subplot(gs[0,2]), 'Transacciones',     f'{len(ventas):,}',          'Total en el año',      '#3BAB6F')
tarjeta(fig.add_subplot(gs[0,3]), 'Unidades Vendidas', f'{unidades_vendidas:,}',    'Artículos despachados','#8E44AD')

# Barras mensuales
ax_bar = fig.add_subplot(gs[1,:])
bars = ax_bar.bar(meses_labels, ventas_mensuales.values, color=CP, edgecolor='white', lw=0.5)
max_idx = list(ventas_mensuales.values).index(max(ventas_mensuales.values))
bars[max_idx].set_color(CA)  # resaltar mejor mes
ax_bar.set_title('Ingresos Mensuales 2025  (naranja = mejor mes)', fontweight='bold', fontsize=13)
ax_bar.set_ylabel('Ingresos ($)', fontsize=10)
ax_bar.yaxis.set_major_formatter(ticker.FuncFormatter(lambda x,_: f'${x:,.0f}'))
for b in bars:
    ax_bar.text(b.get_x()+b.get_width()/2, b.get_height()+500,
                f'${b.get_height()/1000:.0f}K', ha='center', va='bottom', fontsize=7.5, fontweight='bold')

# Día de semana
ax_sem = fig.add_subplot(gs[2,:2])
bars2 = ax_sem.bar(ventas_semana.index, ventas_semana.values, color=COLORES[:7], edgecolor='white', lw=0.5)
ax_sem.set_title('Ingresos por Día de Semana', fontweight='bold', fontsize=13)
ax_sem.set_ylabel('Ingresos ($)', fontsize=10)
ax_sem.yaxis.set_major_formatter(ticker.FuncFormatter(lambda x,_: f'${x:,.0f}'))
for b in bars2:
    ax_sem.text(b.get_x()+b.get_width()/2, b.get_height()+200,
                f'${b.get_height()/1000:.1f}K', ha='center', va='bottom', fontsize=7.5, fontweight='bold')

# Pie categorías
ax_pie = fig.add_subplot(gs[2,2:])
wedges,_,autos = ax_pie.pie(ing_categoria.values, autopct='%1.1f%%',
    colors=COLORES[:len(ing_categoria)], startangle=90, pctdistance=0.80,
    wedgeprops={'edgecolor':'white','linewidth':1.3})
for at in autos: at.set_fontsize(8)
ax_pie.set_title('Ingresos por Categoría', fontweight='bold', fontsize=13)
ax_pie.legend(ing_categoria.index, loc='center left', bbox_to_anchor=(1,0.5), fontsize=8)

plt.savefig('dashboard_ejecutivo.png', dpi=150, bbox_inches='tight')
plt.show()
print(' Dashboard Ejecutivo generado')
```

### 3.5 Dashboard 2 — Análisis de Ventas y Clientes

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 6))
fig.suptitle('Virvuto — Análisis de Ventas  |  Año 2025',
             fontsize=16, fontweight='bold', color='#1A1A2E')

for ax, data, titulo, cols in [
    (axes[0], top_productos.sort_values(), 'Top 5 Productos por Ingresos', COLORES[:5]),
    (axes[1], top_clientes.sort_values(),  'Top 5 Clientes por Ingresos',  COLORES[5:10]),
]:
    bars = ax.barh(range(len(data)), data.values, color=cols, edgecolor='white', height=0.6)
    ax.set_yticks(range(len(data)))
    ax.set_yticklabels([t[:26] for t in data.index], fontsize=10)
    ax.set_title(titulo, fontweight='bold', fontsize=12)
    ax.set_xlabel('Ingresos ($)', fontsize=10)
    ax.xaxis.set_major_formatter(ticker.FuncFormatter(lambda x,_: f'${x:,.0f}'))
    for b in bars:
        ax.text(b.get_width()+50, b.get_y()+b.get_height()/2,
                f'${b.get_width():,.0f}', va='center', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.savefig('dashboard_ventas.png', dpi=150, bbox_inches='tight')
plt.show()
print('Dashboard de Ventas generado')
```
